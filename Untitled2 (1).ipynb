````python
def inspect_dataset(df):

    print("===== DATASET OVERVIEW =====")

    # Dataset dimensions
    print("Rows:", df.shape[0])
    print("Columns:", df.shape[1])

    # Column information
    print("\n===== COLUMN INFORMATION =====")
    print(df.dtypes)

    # Separate numerical and categorical columns
    numerical_cols = df.select_dtypes(include=np.number).columns.tolist()

    categorical_cols = df.select_dtypes(
        include=["object", "category"]
    ).columns.tolist()

    print("\nNumerical Columns:", numerical_cols)
    print("\nCategorical Columns:", categorical_cols)

    # Data quality
    print("\n===== DATA QUALITY =====")
    print("Missing Values:", df.isnull().sum().sum())
    print("Duplicate Rows:", df.duplicated().sum())

    # Statistical summary
    print("\n===== STATISTICAL SUMMARY =====")
    display(df.describe(include="all"))

    return {
        "rows": df.shape[0],
        "columns": df.shape[1],
        "numerical_columns": numerical_cols,
        "categorical_columns": categorical_cols
    }


dataset_info = inspect_dataset(df)
````

````python
def calculate_statistics(df):

    # Select numerical columns
    numerical_data = df.select_dtypes(include=np.number)

    # Calculate descriptive statistics
    if numerical_data.empty:
        return "No numerical columns available for analysis."

    statistics = numerical_data.agg(
        ["mean", "median", "std", "min", "max"]
    )

    return statistics


stats = calculate_statistics(df)

display(stats)
````

````python
def analyze_categorical(df):

    # Identify categorical columns
    categorical_data = df.select_dtypes(
        include=["object", "category"]
    )

    if categorical_data.empty:
        return "No categorical columns available."

    results = {}

    for column in categorical_data.columns:

        results[column] = {
            "unique_values": categorical_data[column].nunique(),
            "most_frequent": categorical_data[column].mode().iloc[0],
            "frequency": categorical_data[column].value_counts().head(5)
        }

    return results


categorical_stats = analyze_categorical(df)

for column, result in categorical_stats.items():

    print("\nColumn:", column)
    print("Unique Values:", result["unique_values"])
    print("Most Frequent:", result["most_frequent"])
    print("Top 5 Frequencies:")
    display(result["frequency"])
````

````python
def analyze_dates(df):

    # Identify possible date columns
    date_columns = []

    for column in df.columns:

        if "date" in column.lower():
            date_columns.append(column)

    if not date_columns:
        return "No date columns found."

    results = {}

    for column in date_columns:

        # Convert column into datetime format
        dates = pd.to_datetime(df[column], errors="coerce")

        results[column] = {
            "earliest_date": dates.min(),
            "latest_date": dates.max(),
            "valid_dates": dates.notna().sum(),
            "invalid_dates": dates.isna().sum()
        }

    return results


date_stats = analyze_dates(df)

for column, result in date_stats.items():

    print("\nColumn:", column)
    print("Earliest Date:", result["earliest_date"])
    print("Latest Date:", result["latest_date"])
    print("Valid Dates:", result["valid_dates"])
    print("Invalid Dates:", result["invalid_dates"])
````

````python
def filter_data(df, column, operator, value):

    # Check whether column exists
    if column not in df.columns:
        return f"Error: Column '{column}' does not exist."

    # Approved filtering operations
    if operator == "equals":

        if pd.api.types.is_string_dtype(df[column]):
            result = df[
                df[column].astype("string").str.strip().str.lower()
                == str(value).strip().lower()
            ]
        else:
            result = df[df[column] == value]

    elif operator == "greater_than":
        result = df[df[column] > value]

    elif operator == "less_than":
        result = df[df[column] < value]

    elif operator == "contains":
        result = df[
            df[column].astype("string").str.contains(
                str(value), case=False, na=False, regex=False
            )
        ]

    elif operator == "between":

        if not isinstance(value, (list, tuple)) or len(value) != 2:
            return "Error: Between requires two values."

        lower, upper = value

        if lower > upper:
            return "Error: Lower value cannot exceed upper value."

        result = df[df[column].between(lower, upper)]

    else:
        return f"Error: Unsupported operator '{operator}'."

    return result




````

````python
def group_analysis(df, group_column, target_column, operation):

    # Validate column names
    if group_column not in df.columns:
        return f"Column '{group_column}' does not exist."

    if target_column not in df.columns:
        return f"Column '{target_column}' does not exist."

    # Approved aggregation operations
    if operation not in ["sum", "mean", "median", "min", "max", "count"]:
        return "Unsupported aggregation operation."

    # Perform grouping and aggregation
    if operation == "count":
        result = df.groupby(group_column)[target_column].count()
    else:
        if not pd.api.types.is_numeric_dtype(df[target_column]):
            return "This aggregation requires a numerical column."

        result = df.groupby(group_column)[target_column].agg(operation)

    return result.sort_values(ascending=False)
````

````python
def generate_plot(df, chart_type, x_column, y_column=None):

    # Validate x column
    if x_column not in df.columns:
        return f"Column '{x_column}' does not exist."

    # Validate y column if provided
    if y_column is not None and y_column not in df.columns:
        return f"Column '{y_column}' does not exist."

    plt.figure(figsize=(10, 5))

    # 1. Histogram
    if chart_type == "histogram":

        if not pd.api.types.is_numeric_dtype(df[x_column]):
            return "Histogram requires a numerical column."

        df[x_column].dropna().plot(kind="hist", bins=15)
        plt.xlabel(x_column)
        plt.ylabel("Frequency")
        plt.title(f"Distribution of {x_column}")

    # 2. Bar chart
    elif chart_type == "bar":

        if y_column is None:
            counts = df[x_column].value_counts().head(15)
            counts.plot(kind="bar")
            plt.ylabel("Count")
        else:
            if not pd.api.types.is_numeric_dtype(df[y_column]):
                return "Bar chart requires a numerical y column."

            df.groupby(x_column)[y_column].mean().sort_values(
                ascending=False
            ).head(15).plot(kind="bar")
            plt.ylabel(f"Average {y_column}")

        plt.xlabel(x_column)
        plt.title(f"Bar Chart: {x_column}")

    # 3. Scatter plot
    elif chart_type == "scatter":

        if y_column is None:
            return "Scatter plot requires both x and y columns."

        if not pd.api.types.is_numeric_dtype(df[x_column]):
            return "Scatter plot requires numerical x and y columns."

        if not pd.api.types.is_numeric_dtype(df[y_column]):
            return "Scatter plot requires numerical x and y columns."

        plt.scatter(df[x_column], df[y_column], alpha=0.7)
        plt.xlabel(x_column)
        plt.ylabel(y_column)
        plt.title(f"{x_column} vs {y_column}")

    # 4. Box plot
    elif chart_type == "box":

        if not pd.api.types.is_numeric_dtype(df[x_column]):
            return "Box plot requires a numerical column."

        df.boxplot(column=x_column)
        plt.ylabel(x_column)
        plt.title(f"Box Plot: {x_column}")

    else:
        return "Unsupported chart type."

    plt.tight_layout()
    plt.show()

    return "Plot generated successfully."
````

````python
from langchain_core.tools import tool
from typing import Any

@tool
def dataset_inspection_tool() -> str:
    """Inspect dataset dimensions, columns, data types,
    missing values, duplicates and statistical summary."""
    return str(inspect_dataset(df))


@tool
def categorical_analysis_tool() -> str:
    """Analyze categorical columns, unique values,
    most frequent values and category frequencies."""
    return str(analyze_categorical(df))


@tool
def date_analysis_tool() -> str:
    """Analyze date columns and identify earliest,
    latest, valid and invalid dates."""
    return str(analyze_dates(df))


@tool
def filter_data_tool(
    column: str,
    operator: str,
    value: Any
) -> str:
    """Filter dataset rows using equals, greater_than,
    less_than, contains or between operators."""
    result = filter_data(df, column, operator, value)
    return result.to_string() if hasattr(result, "to_string") else str(result)


@tool
def group_analysis_tool(
    group_column: str,
    target_column: str,
    operation: str
) -> str:
    """Group dataset by a column and calculate sum, mean,
    median, min, max or count on another column."""
    result = group_analysis(df, group_column, target_column, operation)
    return result.to_string() if hasattr(result, "to_string") else str(result)


@tool
def visualization_tool(
    chart_type: str,
    x_column: str,
    y_column: str = None
) -> str:
    """Generate histogram, bar, scatter or box plots
    using the dataset."""
    generate_plot(df, chart_type, x_column, y_column)
    return "Visualization generated successfully."


# Register all tools together
tools = [
    statistics_tool,
    dataset_inspection_tool,
    categorical_analysis_tool,
    date_analysis_tool,
    filter_data_tool,
    group_analysis_tool,
    visualization_tool
]

print("Total registered tools:", len(tools))

for selected_tool in tools:
    print("-", selected_tool.name)
````

````python
import json
import torch

def select_tool(user_question):

    system_prompt = f"""
You are an AI data analysis assistant.

You have access to the following tools:

{json.dumps(tool_descriptions, indent=2)}

Your task:
1. Understand the user's question.
2. Select the most appropriate tool.
3. Provide the required arguments.
4. Return ONLY valid JSON.

Use this exact JSON format:

{{
    "tool": "tool_name",
    "arguments": {{
        "argument_name": "value"
    }}
}}

If no tool is appropriate, return:

{{
    "tool": "none",
    "arguments": {{}}
}}

Do not execute calculations yourself.
Only select the appropriate tool.
"""

    messages = [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_question}
    ]

    prompt = tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True
    )

    inputs = tokenizer(
        prompt,
        return_tensors="pt"
    ).to(model.device)

    with torch.inference_mode():
        outputs = model.generate(
            **inputs,
            max_new_tokens=250,
            do_sample=False
        )

    response = tokenizer.decode(
        outputs[0][inputs["input_ids"].shape[1]:],
        skip_special_tokens=True
    )

    return response
````

````python

import json
import re
import torch

# --------------------------------------------------
# 1. CREATE A TOOL LOOKUP
# --------------------------------------------------

tool_registry = {
    selected_tool.name: selected_tool
    for selected_tool in tools
}


# --------------------------------------------------
# 2. HELPER FUNCTION: GENERATE QWEN RESPONSE
# --------------------------------------------------

def generate_qwen_response(messages, max_new_tokens=300):

    prompt = tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True
    )

    inputs = tokenizer(
        prompt,
        return_tensors="pt"
    ).to(model.device)

    with torch.inference_mode():
        outputs = model.generate(
            **inputs,
            max_new_tokens=max_new_tokens,
            do_sample=False
        )

    generated_tokens = outputs[0][
        inputs["input_ids"].shape[1]:
    ]

    response = tokenizer.decode(
        generated_tokens,
        skip_special_tokens=True
    )

    return response.strip()


# --------------------------------------------------
# 3. PARSE QWEN'S TOOL SELECTION
# --------------------------------------------------

def parse_tool_decision(response):

    # Remove Markdown code fences if Qwen adds them
    response = re.sub(
        r"```(?:json)?|```",
        "",
        response
    ).strip()

    # Extract JSON object
    start = response.find("{")
    end = response.rfind("}")

    if start == -1 or end == -1:
        raise ValueError("Qwen did not return valid JSON.")

    decision = json.loads(response[start:end + 1])

    return decision


# --------------------------------------------------
# 4. VALIDATE AND NORMALIZE TOOL ARGUMENTS
# --------------------------------------------------


def validate_tool_arguments(tool_name, arguments):

    if not isinstance(arguments, dict):
        raise ValueError("Arguments must be a dictionary.")

    if tool_name not in tool_registry:
        raise ValueError("Tool is not registered.")

    selected_tool = tool_registry[tool_name]

    # Check allowed arguments
    allowed_arguments = set(selected_tool.args.keys())

    unexpected = set(arguments.keys()) - allowed_arguments

    if unexpected:
        raise ValueError(f"Unexpected arguments: {unexpected}")

    # Normalize column names
    column_arguments = [
        "column",
        "group_column",
        "target_column",
        "x_column",
        "y_column"
    ]

    def normalize_name(name):
        return re.sub(r"[^a-z0-9]", "", str(name).lower())

    column_lookup = {
        normalize_name(column): column
        for column in df.columns
    }

    for argument_name in column_arguments:

        if argument_name in arguments:

            value = arguments[argument_name]

            if value is None:
                continue

            normalized_value = normalize_name(value)

            if normalized_value not in column_lookup:
                raise ValueError(
                    f"Column '{value}' does not exist."
                )

            arguments[argument_name] = column_lookup[
                normalized_value
            ]

    # Validate aggregation operations
    if "operation" in arguments:

        allowed_operations = [
            "sum", "mean", "median",
            "min", "max", "count"
        ]

        if arguments["operation"] not in allowed_operations:
            raise ValueError("Unsupported aggregation operation.")

    # Validate filtering operators
    if "operator" in arguments:

        allowed_operators = [
            "equals", "greater_than",
            "less_than", "contains", "between"
        ]

        if arguments["operator"] not in allowed_operators:
            raise ValueError("Unsupported filtering operator.")

    # Validate chart types
    if "chart_type" in arguments:

        allowed_charts = [
            "histogram", "bar", "scatter", "box"
        ]

        if arguments["chart_type"] not in allowed_charts:
            raise ValueError("Unsupported chart type.")

    return arguments


    # Restrict aggregation operations
    if "operation" in arguments:

        allowed_operations = [
            "sum", "mean", "median",
            "min", "max", "count"
        ]

        if arguments["operation"] not in allowed_operations:
            raise ValueError("Unsupported aggregation operation.")

    # Restrict filtering operators
    if "operator" in arguments:

        allowed_operators = [
            "equals", "greater_than",
            "less_than", "contains", "between"
        ]

        if arguments["operator"] not in allowed_operators:
            raise ValueError("Unsupported filtering operator.")

    # Restrict visualization types
    if "chart_type" in arguments:

        allowed_charts = [
            "histogram", "bar", "scatter", "box"
        ]

        if arguments["chart_type"] not in allowed_charts:
            raise ValueError("Unsupported chart type.")

    return arguments


# --------------------------------------------------
# 5. EXECUTE THE SELECTED LANGCHAIN TOOL
# --------------------------------------------------

def execute_tool(tool_name, arguments):

    if tool_name not in tool_registry:
        raise ValueError("Requested tool is not available.")

    arguments = validate_tool_arguments(
        tool_name,
        arguments
    )

    selected_tool = tool_registry[tool_name]

    result = selected_tool.invoke(arguments)

    return str(result)


# --------------------------------------------------
# 6. COMPLETE AGENT CYCLE
# --------------------------------------------------

def run_agent(user_question):

    print("USER QUESTION:")
    print(user_question)

    # Stage A: Ask Qwen to select a tool

    print("\nQWEN: Selecting an appropriate tool...")

    decision_response = select_tool(user_question)

    decision = parse_tool_decision(decision_response)

    tool_name = decision.get("tool")
    arguments = decision.get("arguments", {})

    if tool_name == "none":
        return generate_qwen_response([
            {
                "role": "system",
                "content": "Answer the user's question clearly."
            },
            {
                "role": "user",
                "content": user_question
            }
        ])

    print("\nSELECTED TOOL:", tool_name)
    print("ARGUMENTS:", arguments)

    # Stage B: Execute selected tool

    print("\nPYTHON: Executing the selected tool...")

    try:
        tool_result = execute_tool(
            tool_name,
            arguments
        )

    except Exception as error:
        return f"Tool execution failed: {error}"

    print("\nTOOL RESULT:")
    print(tool_result)

    # Stage C: Send the tool result back to Qwen

    print("\nQWEN: Interpreting the tool result...")

    final_messages = [
        {
            "role": "system",
            "content": (
                "You are an AI data analysis assistant. "
                "Answer using the supplied tool result. "
                "Do not invent numbers or claim calculations "
                "that are not supported by the result. "
                "Explain the findings clearly."
            )
        },
        {
            "role": "user",
            "content": (
                f"Original question: {user_question}\n\n"
                f"Tool used: {tool_name}\n\n"
                f"Actual tool result:\n{tool_result}\n\n"
                "Now answer the original question using these results."
            )
        }
    ]

    final_answer = generate_qwen_response(
        final_messages,
        max_new_tokens=300
    )

    print("\nFINAL ANSWER:")
    print(final_answer)

    return final_answer


# --------------------------------------------------
# 7. TEST OUR AGENT
# --------------------------------------------------

question = "Which region has the highest total units sold?"

answer = run_agent(question)
````

````python

# ============================================================
# AGENTIC DATA ANALYSIS ASSISTANT - LANGGRAPH BACKEND
# ============================================================

import json
import re
import torch

from typing import TypedDict, Optional
from langgraph.graph import StateGraph, START, END


# ============================================================
# 1. CHECK EXISTING COMPONENTS
# ============================================================

required_variables = [
    "model",
    "tokenizer",
    "df",
    "tools",
    "tool_descriptions",
    "generate_qwen_response"
]

for variable in required_variables:
    if variable not in globals():
        raise RuntimeError(
            f"{variable} is missing. Restore it before building the graph."
        )

tool_registry = {
    selected_tool.name: selected_tool
    for selected_tool in tools
}

print("Registered tools:", list(tool_registry.keys()))


# ============================================================
# 2. DEFINE AGENT STATE
# ============================================================

class AgentState(TypedDict):
    question: str
    tool_name: str
    arguments: dict
    tool_result: str
    tool_history: list
    steps: int
    error: str
    final_answer: str


MAX_TOOL_CALLS = 3


# ============================================================
# 3. TOOL SELECTION NODE
# ============================================================

def select_tool_node(state: AgentState):

    question = state["question"]
    history = state.get("tool_history", [])
    steps = state.get("steps", 0)

    if steps >= MAX_TOOL_CALLS:
        return {
            "tool_name": "none",
            "arguments": {}
        }

    history_text = json.dumps(history, indent=2)

    system_prompt = f"""
You are the tool-selection controller for an AI data analyst.

Available tools:
{json.dumps(tool_descriptions, indent=2)}

Previous tool executions:
{history_text}

Original user question:
{question}

Instructions:

1. Select the most appropriate available tool.
2. Supply exact arguments matching its schema.
3. Use actual dataset column names where possible.
4. Do not invent columns or operations.
5. Do not repeat an identical tool call.
6. If previous results already answer the question,
   return tool = none.
7. If no tool is necessary, return tool = none.
8. Return ONLY valid JSON.

Required format:

{{
    "tool": "tool_name",
    "arguments": {{}}
}}

If no tool is needed:

{{
    "tool": "none",
    "arguments": {{}}
}}
"""

    response = generate_qwen_response(
        [
            {"role": "system", "content": system_prompt},
            {
                "role": "user",
                "content": question
            }
        ],
        max_new_tokens=250
    )

    try:

        response = re.sub(
            r"```(?:json)?|```",
            "",
            response
        ).strip()

        start = response.find("{")
        end = response.rfind("}")

        if start == -1 or end == -1:
            raise ValueError("No JSON object found.")

        decision = json.loads(response[start:end + 1])

        tool_name = decision.get("tool")
        arguments = decision.get("arguments", {})

        if not isinstance(arguments, dict):
            raise ValueError("Arguments must be a dictionary.")

        if tool_name != "none" and tool_name not in tool_registry:
            raise ValueError("Unknown tool selected.")

        # Prevent repeating the same tool with identical arguments
        for previous in history:
            if (
                previous["tool_name"] == tool_name
                and previous["arguments"] == arguments
            ):
                tool_name = "none"
                arguments = {}
                break

        print("\n[LANGGRAPH] TOOL SELECTION")
        print("Selected:", tool_name)
        print("Arguments:", arguments)

        return {
            "tool_name": tool_name,
            "arguments": arguments,
            "error": ""
        }

    except Exception as error:

        return {
            "tool_name": "none",
            "arguments": {},
            "error": f"Tool selection failed: {error}"
        }


# ============================================================
# 4. ARGUMENT VALIDATION
# ============================================================

def validate_arguments(tool_name, arguments):

    if tool_name not in tool_registry:
        raise ValueError("Tool is not registered.")

    selected_tool = tool_registry[tool_name]

    allowed_arguments = set(selected_tool.args.keys())

    if not isinstance(arguments, dict):
        raise ValueError("Arguments must be a dictionary.")

    unexpected = set(arguments.keys()) - allowed_arguments

    if unexpected:
        raise ValueError(
            f"Unexpected arguments: {unexpected}"
        )

    # Normalize column names
    def normalize(value):
        return re.sub(
            r"[^a-z0-9]",
            "",
            str(value).lower()
        )

    column_lookup = {
        normalize(column): column
        for column in df.columns
    }

    column_arguments = [
        "column",
        "group_column",
        "target_column",
        "x_column",
        "y_column"
    ]

    arguments = arguments.copy()

    for argument in column_arguments:

        if argument in arguments:

            value = arguments[argument]

            if value is None:
                continue

            normalized = normalize(value)

            if normalized not in column_lookup:
                raise ValueError(
                    f"Column '{value}' does not exist."
                )

            arguments[argument] = column_lookup[normalized]

    # Validate aggregation operations
    if "operation" in arguments:

        if arguments["operation"] not in [
            "sum", "mean", "median",
            "min", "max", "count"
        ]:
            raise ValueError("Invalid aggregation operation.")

    # Validate filter operators
    if "operator" in arguments:

        if arguments["operator"] not in [
            "equals", "greater_than",
            "less_than", "contains", "between"
        ]:
            raise ValueError("Invalid filter operator.")

    # Validate chart types
    if "chart_type" in arguments:

        if arguments["chart_type"] not in [
            "histogram", "bar", "scatter", "box"
        ]:
            raise ValueError("Invalid chart type.")

    return arguments


# ============================================================
# 5. TOOL EXECUTION NODE
# ============================================================

def execute_tool_node(state: AgentState):

    tool_name = state["tool_name"]
    arguments = state["arguments"]

    try:

        validated_arguments = validate_arguments(
            tool_name,
            arguments
        )

        print("\n[LANGGRAPH] EXECUTING TOOL:", tool_name)

        result = tool_registry[tool_name].invoke(
            validated_arguments
        )

        result = str(result)

        history = state.get("tool_history", [])

        new_history = history + [
            {
                "tool_name": tool_name,
                "arguments": validated_arguments,
                "result": result[:12000]
            }
        ]

        print("[LANGGRAPH] TOOL EXECUTED SUCCESSFULLY")

        return {
            "tool_result": result,
            "tool_history": new_history,
            "steps": state.get("steps", 0) + 1,
            "error": ""
        }

    except Exception as error:

        print("[LANGGRAPH] TOOL ERROR:", error)

        return {
            "error": str(error),
            "steps": state.get("steps", 0) + 1
        }


# ============================================================
# 6. FINAL ANSWER NODE
# ============================================================

def final_answer_node(state: AgentState):

    question = state["question"]
    history = state.get("tool_history", [])
    error = state.get("error", "")

    if error and not history:

        return {
            "final_answer": (
                "I could not complete the analysis. "
                f"Reason: {error}"
            )
        }

    if not history:

        answer = generate_qwen_response(
            [
                {
                    "role": "system",
                    "content": (
                        "You are a helpful AI data analysis assistant. "
                        "Answer general questions clearly. "
                        "Do not invent dataset results."
                    )
                },
                {
                    "role": "user",
                    "content": question
                }
            ],
            max_new_tokens=300
        )

        return {"final_answer": answer}

    result_context = json.dumps(
        history,
        indent=2
    )

    if error:
        result_context += "\nExecution warning: " + error

    messages = [
        {
            "role": "system",
            "content": (
                "You are an AI data analyst. "
                "Explain the actual tool results clearly. "
                "Use only the supplied results. "
                "Never invent numerical values. "
                "If an execution failed, clearly mention it."
            )
        },
        {
            "role": "user",
            "content": (
                f"User question: {question}\n\n"
                f"Analysis results:\n{result_context}\n\n"
                "Provide the final answer."
            )
        }
    ]

    answer = generate_qwen_response(
        messages,
        max_new_tokens=350
    )

    print("\n[LANGGRAPH] FINAL ANSWER GENERATED")

    return {
        "final_answer": answer
    }


# ============================================================
# 7. CONDITIONAL ROUTING
# ============================================================

def route_after_selection(state: AgentState):

    if state.get("error"):
        return "final"

    if state.get("tool_name") == "none":
        return "final"

    return "execute"


def route_after_execution(state: AgentState):

    if state.get("error"):
        return "final"

    if state.get("steps", 0) >= MAX_TOOL_CALLS:
        return "final"

    return "select"


# ============================================================
# 8. BUILD LANGGRAPH
# ============================================================

workflow = StateGraph(AgentState)

# Add nodes
workflow.add_node("select", select_tool_node)
workflow.add_node("execute", execute_tool_node)
workflow.add_node("final", final_answer_node)

# Starting edge
workflow.add_edge(START, "select")

# Conditional routing after tool selection
workflow.add_conditional_edges(
    "select",
    route_after_selection,
    {
        "execute": "execute",
        "final": "final"
    }
)

# After execution, either select another tool or finish
workflow.add_conditional_edges(
    "execute",
    route_after_execution,
    {
        "select": "select",
        "final": "final"
    }
)

# Final edge
workflow.add_edge("final", END)

# Compile the graph
agent_graph = workflow.compile()

print("\nLANGGRAPH CREATED SUCCESSFULLY!")
print("Nodes: select, execute, final")
print("Maximum tool calls:", MAX_TOOL_CALLS)


# ============================================================
# 9. CREATE A USER-FACING FUNCTION
# ============================================================

def ask_data_agent(question):

    initial_state = {
        "question": question,
        "tool_name": "none",
        "arguments": {},
        "tool_result": "",
        "tool_history": [],
        "steps": 0,
        "error": "",
        "final_answer": ""
    }

    result = agent_graph.invoke(initial_state)

    return result["final_answer"]


# ============================================================
# 10. TEST THE COMPLETE LANGGRAPH AGENT
# ============================================================

test_question = (
    "Which region has the highest total units sold?"
)

final_response = ask_data_agent(test_question)

print("\n======================================")
print("USER QUESTION:")
print(test_question)

print("\nFINAL AI RESPONSE:")
print(final_response)

print("======================================")
````

````python

# ============================================================
# AGENTIC DATA ANALYSIS ASSISTANT
# FASTAPI + NGROK BACKEND
# ============================================================

# 1. Install required packages
%pip install -q fastapi uvicorn pyngrok python-multipart nest-asyncio

# 2. Import libraries
import io
import os
import time
import secrets
import threading
import pandas as pd
import uvicorn

from fastapi import FastAPI, UploadFile, File, HTTPException, Header, Depends
from fastapi.middleware.cors import CORSMiddleware
from pydantic import BaseModel
from starlette.concurrency import run_in_threadpool
from pyngrok import ngrok

# 3. Verify existing project components
if "df" not in globals():
    raise RuntimeError("DataFrame 'df' not found. Load your CSV first.")

if "ask_data_agent" not in globals():
    raise RuntimeError("LangGraph function 'ask_data_agent' not found.")





# 5. Create FastAPI application
app = FastAPI(
    title="Agentic Data Analysis Assistant",
    description="Qwen + LangChain + LangGraph + Pandas",
    version="1.0"
)

# CORS configuration for initial demo
app.add_middleware(
    CORSMiddleware,
    allow_origins=["*"],
    allow_credentials=False,
    allow_methods=["*"],
    allow_headers=["*"]
)

# 6. API authentication
API_KEY = secrets.token_urlsafe(32)

def verify_api_key(authorization: str = Header(None)):
    if authorization != f"Bearer {API_KEY}":
        raise HTTPException(
            status_code=401,
            detail="Invalid or missing API key"
        )
    return True

# 7. Lock to prevent simultaneous model inference
agent_lock = threading.Lock()

def run_agent_safely(question):
    with agent_lock:
        return ask_data_agent(question)

# 8. Request model for chat
class ChatRequest(BaseModel):
    question: str

# 9. Health-check endpoint
@app.get("/health")
def health_check():
    return {
        "status": "online",
        "message": "Agentic Data Analysis Backend is running!"
    }

# 10. CSV upload endpoint
@app.post("/upload")
async def upload_csv(
    file: UploadFile = File(...),
    authenticated: bool = Depends(verify_api_key)
):
    global df

    if not file.filename.lower().endswith(".csv"):
        raise HTTPException(
            status_code=400,
            detail="Please upload a CSV file."
        )

    try:
        contents = await file.read()

        if not contents:
            raise ValueError("Uploaded file is empty.")

        new_df = pd.read_csv(io.BytesIO(contents))

        if new_df.empty:
            raise ValueError("CSV contains no data.")

        # Update the active dataset used by existing tools
        df = new_df

        return {
            "status": "success",
            "message": "Dataset uploaded successfully!",
            "filename": file.filename,
            "rows": int(df.shape[0]),
            "columns": int(df.shape[1]),
            "column_names": df.columns.tolist()
        }

    except Exception as e:
        raise HTTPException(
            status_code=400,
            detail=f"CSV upload failed: {str(e)}"
        )

# 11. Chat endpoint
@app.post("/chat")
async def chat(
    request: ChatRequest,
    authenticated: bool = Depends(verify_api_key)
):
    question = request.question.strip()

    if not question:
        raise HTTPException(
            status_code=400,
            detail="Question cannot be empty."
        )

    try:
        result = await run_in_threadpool(
            run_agent_safely,
            question
        )

        # Extract final answer from LangGraph state
        if isinstance(result, dict):
            answer = result.get(
                "final_answer",
                str(result)
            )
            steps = result.get("steps", [])
            tool_history = result.get("tool_history", [])
        else:
            answer = str(result)
            steps = []
            tool_history = []

        return {
            "status": "success",
            "question": question,
            "answer": answer,
            "steps": steps,
            "tool_history": tool_history
        }

    except Exception as e:
        raise HTTPException(
            status_code=500,
            detail=f"Agent execution failed: {str(e)}"
        )

# 12. Start Uvicorn server
config = uvicorn.Config(
    app,
    host="0.0.0.0",
    port=8000,
    log_level="info"
)

server = uvicorn.Server(config)

server_thread = threading.Thread(
    target=server.run,
    daemon=True
)

server_thread.start()

# Wait for server startup
for attempt in range(30):
    if server.started:
        break
    time.sleep(1)

if not server.started:
    raise RuntimeError("FastAPI server failed to start.")

# 13. Create public Ngrok tunnel
ngrok.kill()

public_tunnel = ngrok.connect(8000)
BACKEND_URL = public_tunnel.public_url

# 14. Display connection information
print("=" * 55)
print("AGENTIC DATA ANALYSIS BACKEND IS ONLINE")
print("=" * 55)

print("\nBackend URL:", BACKEND_URL)
print("Health URL:", BACKEND_URL + "/health")
print("API documentation:", BACKEND_URL + "/docs")

print("\nAPI KEY (copy privately into frontend notebook):")
print(API_KEY)

print("\nAvailable endpoints:")
print("GET  /health")
print("POST /upload")
print("POST /chat")

print("\nBackend is ready!")
````
